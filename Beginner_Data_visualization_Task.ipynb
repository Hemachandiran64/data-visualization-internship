###Visualization Library Documentation

**Matplotlib and Seaborn**

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

print("Matplotlib and Seaborn are ready!")
```

```python
data = {
    "Month": ["Jan", "Feb", "Mar", "Apr", "May"],
    "Sales": [120, 150, 180, 140, 200],
    "Students": [20, 25, 30, 28, 35]
}

df = pd.DataFrame(data)

df
```

### 1. Line Plot

A line plot is used to show changes or trends over time.

Use case:
It can be used to show monthly sales trends.

```python
plt.plot(df["Month"], df["Sales"], marker="o")

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

### 2. Bar Chart

A bar chart is used to compare values between different categories.

Use case:
It can be used to compare sales between months.

```python
plt.bar(df["Month"], df["Sales"])

plt.title("Monthly Sales")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

  ### 3. Scatter Plot

A scatter plot is used to show the relationship between two numerical variables.

Use case:
It can be used to study the relationship between students and sales.

```python
plt.scatter(df["Students"], df["Sales"])

plt.title("Students vs Sales")
plt.xlabel("Number of Students")
plt.ylabel("Sales")

plt.show()
```

### 4. Histogram

A histogram is used to show the distribution of numerical data.

Use case:
It can be used to understand how marks are distributed among students.

```python
marks = [45, 50, 55, 60, 60, 65, 70, 70, 75, 80, 85, 90, 95]

plt.hist(marks, bins=5)

plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

### 5. Pie Chart

A pie chart is used to show the proportion or percentage of different categories.

Use case:
It can be used to show the distribution of students according to different categories.

```python
categories = ["Excellent", "Good", "Average", "Needs Improvement"]
students = [10, 15, 8, 5]

plt.pie(students, labels=categories, autopct="%1.1f%%")

plt.title("Student Performance Distribution")

plt.show()
```

## Seaborn

### Library Overview

Seaborn is a Python data visualization library built on top of Matplotlib.

It provides simple and attractive statistical graphics and works well with Pandas DataFrames.

### Typical Use Cases

Seaborn is commonly used for:
- Statistical data visualization
- Exploring relationships between variables
- Understanding data distributions
- Comparing different categories

### 1. Line Plot

A line plot is used to show trends and changes in data.

Use case:
It can be used to show monthly sales trends.

```python
sns.lineplot(data=df, x="Month", y="Sales", marker="o")

plt.title("Monthly Sales Using Seaborn")
plt.xlabel("Month")
plt.ylabel("Sales")



plt.show()
```

### 2. Bar Plot

A bar plot is used to compare values between different categories.

Use case:
It can be used to compare sales across different months.

```python
sns.barplot(data=df, x="Month", y="Sales")

plt.title("Monthly Sales Using Seaborn")
plt.xlabel("Month")
plt.ylabel("Sales")

plt.show()
```

### 3. Scatter Plot

A scatter plot is used to show the relationship between two numerical variables.

Use case:
It can be used to study the relationship between the number of students and sales.

```python
sns.scatterplot(data=df, x="Students", y="Sales")

plt.title("Students vs Sales Using Seaborn")
plt.xlabel("Number of Students")
plt.ylabel("Sales")

plt.show()
```

### Important Parts of the Code

1. sns.scatterplot()
   Creates a scatter plot using Seaborn.

2. data=df
   Uses our Pandas DataFrame.

3. x="Students"
   Uses the Students column for the X-axis.

4. y="Sales"
   Uses the Sales column for the Y-axis.

5. plt.title()
   Adds a title to the graph.

6. plt.xlabel()
   Gives a name to the X-axis.

7. plt.ylabel()
   Gives a name to the Y-axis.

8. plt.show()
   Displays the graph.

### 4. Histogram

A histogram is used to show the distribution of numerical data.

Use case:
It can be used to understand how marks are distributed among students.

```python
marks = [45, 50, 55, 60, 60, 65, 70, 70, 75, 80, 85, 90, 95]

sns.histplot(marks, bins=5)

plt.title("Distribution of Marks Using Seaborn")
plt.xlabel("Marks")
plt.ylabel("Number of Students")

plt.show()
```

### Important Parts of the Code

1. sns.histplot()
   Creates a histogram using Seaborn.

2. marks
   Contains the numerical marks data.

3. bins=5
   Divides the data into 5 groups.

4. plt.title()
   Adds a title to the graph.

5. plt.xlabel()
   Gives a name to the X-axis.

6. plt.ylabel()
   Gives a name to the Y-axis.

7. plt.show()
   Displays the histogram.

### 5. Box Plot

A box plot is used to show the distribution of numerical data and identify possible outliers.

Use case:
It can be used to understand the spread of students' marks.

```python
sns.boxplot(y=marks)

plt.title("Distribution of Marks Using Seaborn")
plt.ylabel("Marks")

plt.show()
```

### Important Parts of the Code

1. sns.boxplot()
   Creates a box plot using Seaborn.

2. y=marks
   Uses the marks data for the box plot.

3. plt.title()
   Adds a title to the graph.

4. plt.ylabel()
   Gives a name to the Y-axis.

5. plt.show()
   Displays the box plot.

## Matplotlib vs Seaborn

### Comparison

| Feature | Matplotlib | Seaborn |
|---|---|---|
| Ease of Use | Requires more code and configuration | Easier for statistical plots |
| Customization | Highly customizable | Good customization |
| Interactivity | Mainly static plots | Mainly static plots |
| Statistical Visualization | Basic | Strong support |
| DataFrame Support | Good | Very good |
| Large Datasets | Can handle large datasets, but plotting can become slower | Can also handle large datasets, but performance depends on the plot and data size |

### Matplotlib Strengths

- Highly customizable
- Provides control over almost every part of a graph
- Suitable for creating basic and advanced visualizations

### Matplotlib Weaknesses

- Can require more code
- Some graphs need more manual configuration

### Seaborn Strengths

- Simple syntax
- Attractive statistical visualizations
- Works well with Pandas DataFrames
- Useful for exploring relationships and distributions

### Seaborn Weaknesses

- Provides less low-level control than Matplotlib
- Built mainly for statistical visualization

## Conclusion

Matplotlib and Seaborn are useful Python libraries for data visualization.

Matplotlib provides detailed control and customization of graphs, while Seaborn makes statistical visualization easier and simpler.

Both libraries are useful for Data Science and can help users understand patterns, relationships, and distributions in data.
